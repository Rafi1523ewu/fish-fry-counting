```python
%%bash
# One script, so the first failing step stops the run instead of being skipped over.
set -euo pipefail
git clone -q https://github.com/Rafi1523ewu/fish-fry-counting.git /kaggle/working/fish-fry-counting
pip install -q -e /kaggle/working/fish-fry-counting
# Downloads FFD.zip from Mendeley Data and refuses it unless the SHA-256 matches.
# Images live in /tmp so they are not saved as notebook output.
fishfry prepare --out /tmp/ffd
fishfry grid --data /tmp/ffd --runs /kaggle/working/runs --gpus 2
fishfry summarize --runs /kaggle/working/runs --out /kaggle/working/results
cat /kaggle/working/results/summary.csv /kaggle/working/results/selected_config.txt
```
